```python
import pandas as pd

df = pd.read_csv("bank-full.csv", sep=";")

df["y"] = df["y"].map({"yes": 1, "no": 0})

print("Rows:", len(df))
print("Columns:", len(df.columns))

df.head()
```

```python
# Part 1: Base rate

base_rate = df['y'].mean()

expected_yes_10000 = base_rate * 10000

print(f"Base rate: {base_rate:.4f} ({base_rate*100:.2f}%)")
print(f"Expected yeses from 10,000 random calls: {expected_yes_10000:.0f}")
```

```python
import numpy as np
import pandas as pd
```

```python
# Part 2: Response rate by call duration

bins = [0, 60, 120, 240, 480, np.inf]
labels = ['under 1 min', '1–2 min', '2–4 min', '4–8 min', '8+ min']

df['duration_band'] = pd.cut(
    df['duration'],
    bins=bins,
    labels=labels,
    right=False
)

duration_table = (
    df.groupby('duration_band', observed=False)['y']
      .agg(['count', 'sum', 'mean'])
      .reset_index()
)

duration_table['response_rate_%'] = duration_table['mean'] * 100

duration_table
```

```python
# PART 3: Build the propensity model

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Variables known BEFORE the call
features = [
    'age',
    'job',
    'marital',
    'education',
    'default',
    'balance',
    'housing',
    'loan',
    'pdays',
    'previous',
    'poutcome'
]

X = df[features]
y = df['y']

numeric_cols = [
    'age',
    'balance',
    'pdays',
    'previous'
]

categorical_cols = [
    'job',
    'marital',
    'education',
    'default',
    'housing',
    'loan',
    'poutcome'
]

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    stratify=y,
    random_state=42
)

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ]
)

# Logistic regression model
model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', LogisticRegression(max_iter=3000))
    ]
)

# Train
model.fit(X_train, y_train)

# Predict probability of "yes"
scores = model.predict_proba(X_test)[:, 1]

# Create test dataset
test = df.loc[X_test.index].copy()
test['score'] = scores

# Show top 10 customers
top10 = test.sort_values(
    'score',
    ascending=False
).head(10)

top10[
    [
        'age',
        'job',
        'marital',
        'education',
        'balance',
        'housing',
        'loan',
        'pdays',
        'previous',
        'poutcome',
        'y',
        'score'
    ]
]
```

```python
# PART 4: Accuracy trap

from sklearn.metrics import accuracy_score, roc_auc_score

# Model prediction using 0.5 threshold
predictions = (scores >= 0.5).astype(int)

model_accuracy = accuracy_score(
    y_test,
    predictions
)

# Lazy model: always predicts "no"
lazy_predictions = np.zeros(len(y_test))

lazy_accuracy = accuracy_score(
    y_test,
    lazy_predictions
)

# AUC measures ranking quality
auc = roc_auc_score(
    y_test,
    scores
)

print(f"Model accuracy: {model_accuracy:.4f}")
print(f"Lazy 'always no' accuracy: {lazy_accuracy:.4f}")
print(f"AUC: {auc:.4f}")
```

```python
# PART 5: Create 10 customer groups

# Sort customers by predicted probability
test_sorted = test.sort_values(
    'score',
    ascending=False
).copy()

# Create 10 equally sized groups
test_sorted['group'] = pd.qcut(
    test_sorted['score'].rank(
        method='first',
        ascending=False
    ),
    10,
    labels=range(1, 11)
)

# Calculate customers and yeses
group_table = (
    test_sorted
    .groupby('group', observed=False)
    .agg(
        clients=('y', 'size'),
        yeses=('y', 'sum')
    )
)

# Response rate
group_table['yeses_per_100'] = (
    group_table['yeses']
    / group_table['clients']
    * 100
)

# Lift
base_rate = df['y'].mean()

group_table['lift'] = (
    group_table['yeses_per_100']
    / (base_rate * 100)
)

group_table
```

```python
# PART 6: Profit and stopping point

call_cost = 10
deposit_value = 70

# Break-even response rate
break_even = call_cost / deposit_value

print(
    f"Break-even response rate: {break_even * 100:.2f}%"
)

# Profit for each group
group_table['profit'] = (
    group_table['yeses'] * deposit_value
    - group_table['clients'] * call_cost
)

# Running profit
group_table['running_profit'] = (
    group_table['profit'].cumsum()
)

group_table[
    [
        'clients',
        'yeses',
        'yeses_per_100',
        'lift',
        'profit',
        'running_profit'
    ]
]
```

```python
# PART 7: Customer persona for Group 1

group1 = test_sorted[test_sorted['group'] == 1].copy()

print("Number and spread of numeric variables:")
display(group1[['age', 'balance']].describe())

print("\nJob distribution:")
display(
    group1['job']
    .value_counts(normalize=True)
    .mul(100)
    .round(1)
)

print("\nHousing distribution:")
display(
    group1['housing']
    .value_counts(normalize=True)
    .mul(100)
    .round(1)
)

print("\nPrevious campaign outcome:")
display(
    group1['poutcome']
    .value_counts(normalize=True)
    .mul(100)
    .round(1)
)
```

```python
# Persona summary table

persona_df = group1.copy()

main_jobs = ['retired', 'management', 'student']

persona_df['type'] = np.where(
    persona_df['job'].isin(main_jobs),
    persona_df['job'],
    'other jobs'
)

persona_table = (
    persona_df
    .groupby('type')
    .agg(
        clients=('y', 'size'),
        median_age=('age', 'median'),
        median_balance=('balance', 'median'),
        housing_loan_share=(
            'housing',
            lambda x: (x == 'yes').mean() * 100
        ),
        yes_rate=('y', 'mean')
    )
)

persona_table['yes_rate_%'] = (
    persona_table['yes_rate'] * 100
)

persona_table[
    [
        'clients',
        'median_age',
        'median_balance',
        'housing_loan_share',
        'yes_rate_%'
    ]
].sort_values(
    'clients',
    ascending=False
)
```

```python
# PART 8: Key recommendation

print("RECOMMENDATION")
print()
print(
    "The bank should target the highest-scoring customers "
    "and mainly call Groups 1 and 2."
)
print(
    "These groups have the strongest response rates and "
    "generate positive profit."
)
print(
    "The bank should stop calling when the response rate "
    "falls below the break-even rate."
)

print()
print("MAIN RISK")
print()
print(
    "The model may systematically favor or exclude certain "
    "customer groups, such as particular age or job groups."
)
print(
    "The predicted probability shows who is more likely to "
    "respond, but it does not prove that calling the customer "
    "causes the response."
)
```
