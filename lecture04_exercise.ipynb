```python
%pip install pandas plotly
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2002 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
%pip install nbformat
```

```python
task1 = df.loc[
    (df['continent'] == 'Europe') &
    (df['year'].isin([2002, 2007]))
]

fig = px.scatter(
    task1,
    x='gdpPercap',
    y='lifeExp',
    color='year',
    hover_name='country',
    title='Life Expectancy vs GDP per Capita in Europe (2002–2007)',
    labels={
        'gdpPercap': 'GDP per Capita',
        'lifeExp': 'Life Expectancy (years)',
        'year': 'Year'
    }
)

fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
# Task 2
# Task 2 — Bubble chart: tell a story

# Use the full 2007 dataset
task2 = df[df['year'] == 2007].copy()

# Highlight Asia, make all other continents grey
task2['group'] = task2['continent'].apply(
    lambda x: 'Asia' if x == 'Asia' else 'Other continents'
)

# Create bubble chart
fig = px.scatter(
    task2,
    x='gdpPercap',
    y='lifeExp',
    size='pop',
    color='group',
    hover_name='country',
    hover_data=['continent', 'pop'],
    log_x=True,
    size_max=60,
    color_discrete_map={
        'Asia': 'blue',
        'Other continents': 'lightgray'
    },
    title='Life Expectancy, GDP and Population — 2007',
    labels={
        'gdpPercap': 'GDP per Capita (log scale)',
        'lifeExp': 'Life Expectancy (years)',
        'pop': 'Population',
        'group': 'Continent'
    }
)

# Add annotation explaining the highlighted group's story
fig.add_annotation(
    x=2500,
    y=55,
    text='Asia: large populations and a wide range<br>'
         'of GDP per capita and life expectancy',
    showarrow=True,
    arrowhead=2,
    ax=100,
    ay=-80
)

# Improve appearance
fig.update_layout(
    legend_title_text='Highlighted Group',
    width=1000,
    height=650
)

fig.show()
```
